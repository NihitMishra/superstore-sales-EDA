# **Key Insights and Business Recommendations**

### 1. Category-wise Profitability
Technology generated the highest total sales ($836,154.03) and profit ($145,454.95). In contrast, Furniture generated $741,999.80 in sales but achieved a profit margin of only 2.49%.

**Business recommendation:** Review Furniture pricing, discounting practices, and product costs to identify opportunities to improve profitability.

### 2. Regional Performance
The West region recorded the highest sales ($725,457.82) and profit ($108,418.45). The Central region generated more sales than the South region but earned less total profit.

**Business recommendation:** Investigate differences in discount levels, product mix, and costs between Central and South to understand the profitability gap.

### 3. Product and Sub-category Performance
The Canon imageCLASS 2200 Advanced Copier was the top product by sales, generating $61,599.82. Copiers recorded the highest total sub-category profit ($55,617.82), while Tables recorded the largest loss (−$17,725.48).

**Business recommendation:** Review the profitability of high-sales products and investigate losses in Tables before changing inventory or pricing strategies.

### 4. Discount and Profitability Patterns
Average profit was negative for the selected discount groups at 30%, 40%, 50%, and 80%. The 50% discount group recorded an average profit of −$310.70 per row.

**Business recommendation:** Review high-discount transactions and evaluate discount policies by product and sub-category. These results show an association, not proof that discounts alone caused losses.

### 5. Monthly Sales Trends
November showed strong sales across multiple years, reaching $118,447.83 in November 2017. Monthly sales varied considerably throughout the four-year period.

**Business recommendation:** Investigate seasonal demand patterns and consider aligning inventory and promotional planning with historically strong sales periods.

### 6. Customer Segment Performance
The Consumer segment generated the highest sales ($1,161,401.35) and profit ($134,119.21). Corporate ranked second, while Home Office had the lowest totals.

**Business recommendation:** Maintain the Consumer segment's performance while exploring profitable growth opportunities in Corporate and Home Office.

## Conclusion
The exploratory analysis highlights differences in profitability across product categories, regions, sub-categories, discount levels, and customer segments. Technology and the West region were strong performers, while Furniture's low profit margin and losses in the Tables sub-category warrant further investigation. These findings can support more informed decisions about pricing, discounting, product performance, and regional strategy.
